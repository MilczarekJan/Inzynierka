```python
from ultralytics import YOLO

# Create a new YOLO model from scratch
# model = YOLO("yolon.yaml")

# Load a pretrained YOLO model (recommended for training)
model = YOLO("yolov8n.pt")

# Train the model using the 'data.yaml' dataset for 20 epochs
results = model.train(data="/mnt/Inzynierka/elementyv8_3klasy/data.yaml", epochs=20, batch=2)

# Evaluate the model's performance on the validation set
results = model.val()

# Perform object detection on an image using the model
results = model("/mnt/Inzynierka/elementyv11_zbior2_3klasy/test/images/pcb129rec1_jpg.rf.885b3f0e7b6d744cda4b1bdf513ab94b.jpg")

# Export the model to ONNX format
success = model.export(format="onnx")
```
